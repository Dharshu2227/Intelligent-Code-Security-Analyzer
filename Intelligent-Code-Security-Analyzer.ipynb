```python
# ============================================
# STEP 2: PREPARE DOCUMENTS FOR RETRIEVAL
# ============================================

documents = []

for item in security_knowledge:
    text = (
        item["category"] + " " +
        item["title"] + " " +
        item["content"] + " " +
        " ".join(item["keywords"])
    )

    documents.append(text)

print("Documents prepared:", len(documents))
```

```python
# ============================================
# STEP 3: CREATE RETRIEVAL SYSTEM
# ============================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

knowledge_vectors = vectorizer.fit_transform(documents)

print("TF-IDF knowledge index created!")
print("Vector shape:", knowledge_vectors.shape)
```

```python
# ============================================
# STEP 4: RETRIEVE RELEVANT KNOWLEDGE
# ============================================

def retrieve_knowledge(query, top_k=3):

    query_vector = vectorizer.transform([query])

    similarity_scores = cosine_similarity(
        query_vector,
        knowledge_vectors
    )[0]

    ranked_indexes = similarity_scores.argsort()[::-1]

    results = []

    for index in ranked_indexes[:top_k]:

        results.append({
            "category": security_knowledge[index]["category"],
            "title": security_knowledge[index]["title"],
            "content": security_knowledge[index]["content"],
            "score": float(similarity_scores[index])
        })

    return results
```

```python
# ============================================
# STEP 7: SECURITY GUIDANCE FUNCTION
# ============================================

def get_security_guidance(issue, top_k=3):

    results = retrieve_knowledge(issue, top_k)

    guidance = []

    for result in results:
        guidance.append({
            "category": result["category"],
            "title": result["title"],
            "recommendation": result["content"],
            "similarity": round(result["score"], 3)
        })

    return guidance


# Test the function
issue = "hardcoded password found in source code"

guidance = get_security_guidance(issue, top_k=3)

print("SECURITY GUIDANCE")
print("=" * 60)

for item in guidance:
    print("Category:", item["category"])
    print("Issue:", item["title"])
    print("Recommendation:", item["recommendation"])
    print("Similarity:", item["similarity"])
    print("-" * 60)
```

```python
# ============================================
# STEP 8: COMBINE ISSUE + RETRIEVED KNOWLEDGE
# ============================================

def analyze_security_issue(issue, severity="MEDIUM"):

    guidance = get_security_guidance(issue, top_k=3)

    analysis = {
        "issue": issue,
        "severity": severity,
        "recommendations": guidance
    }

    return analysis


# Test
result = analyze_security_issue(
    "hardcoded password found in source code",
    severity="HIGH"
)

print("SECURITY ANALYSIS")
print("=" * 60)

print("Issue:", result["issue"])
print("Severity:", result["severity"])

print("\nRelevant Knowledge:")

for item in result["recommendations"]:
    print("\nCategory:", item["category"])
    print("Rule:", item["title"])
    print("Recommendation:", item["recommendation"])
    print("Similarity:", item["similarity"])
```

```python
# ============================================
# STEP 9: CREATE FINAL RISK REPORT
# ============================================

import pandas as pd

# Example security findings
security_findings = [
    {
        "file": "sample.py",
        "line": 26,
        "issue": "Hardcoded password found in source code",
        "severity": "HIGH"
    },
    {
        "file": "sample.py",
        "line": 10,
        "issue": "Potential security issue detected",
        "severity": "MEDIUM"
    }
]

report_rows = []

for finding in security_findings:

    guidance = get_security_guidance(
        finding["issue"],
        top_k=3
    )

    # Get the most relevant recommendation
    if guidance:
        best = guidance[0]

        recommendation = best["recommendation"]
        rule = best["title"]
        similarity = best["similarity"]
    else:
        recommendation = "No security guidance found."
        rule = "N/A"
        similarity = 0

    report_rows.append({
        "File": finding["file"],
        "Line": finding["line"],
        "Issue": finding["issue"],
        "Severity": finding["severity"],
        "Security Rule": rule,
        "Recommendation": recommendation,
        "Similarity": similarity
    })


# Create DataFrame
final_report = pd.DataFrame(report_rows)

print("FINAL RISK REPORT")
print("=" * 80)

display(final_report)
```

```python
# ============================================
# STEP 9B: SAVE FINAL REPORT AS CSV
# ============================================

output_file = "final_risk_report.csv"

final_report.to_csv(
    output_file,
    index=False
)

print("CSV created successfully!")
print("File:", output_file)
```

```python
# ============================================
# STEP 16: RADON COMPLEXITY ANALYSIS
# ============================================

import subprocess
import json

radon_result = subprocess.run(
    ["radon", "cc", "sample.py", "-j"],
    capture_output=True,
    text=True
)

print("Radon output:")
print(radon_result.stdout)
```

```python
# ============================================
# STEP 16B: READ RADON RESULTS
# ============================================

radon_data = json.loads(radon_result.stdout)

print("RADON RESULTS")
print("=" * 60)

for filename, functions in radon_data.items():

    for function in functions:

        print("Function:", function["name"])
        print("Complexity:", function["complexity"])
        print("Rank:", function["rank"])
        print("Line:", function["lineno"])
        print("-" * 60)
```

```python
# ============================================
# STEP 17: CALCULATE RISK SCORE
# ============================================

severity_points = {
    "LOW": 1,
    "MEDIUM": 5,
    "HIGH": 10
}

total_security_points = 0

for finding in findings:
    severity = finding["severity"].upper()
    total_security_points += severity_points.get(severity, 0)

# Calculate average complexity
complexities = []

for filename, functions in radon_data.items():
    for function in functions:
        complexities.append(function["complexity"])

if complexities:
    average_complexity = sum(complexities) / len(complexities)
else:
    average_complexity = 0

# Complexity contribution
if average_complexity <= 5:
    complexity_points = 0
elif average_complexity <= 10:
    complexity_points = 5
else:
    complexity_points = 10

# Final score
risk_score = total_security_points + complexity_points

# Keep score between 0 and 100
risk_score = min(risk_score, 100)

# Risk category
if risk_score >= 20:
    risk_level = "HIGH"
elif risk_score >= 10:
    risk_level = "MEDIUM"
else:
    risk_level = "LOW"

print("RISK ANALYSIS")
print("=" * 60)
print("Security points:", total_security_points)
print("Average complexity:", round(average_complexity, 2))
print("Complexity points:", complexity_points)
print("Final risk score:", risk_score)
print("Risk level:", risk_level)
```

```python
# ============================================
# STEP 18: GENERATE FINAL RISK REPORT
# ============================================

final_rows = []

for finding in findings:

    guidance = get_security_guidance(
        finding["issue"],
        top_k=3
    )

    best = guidance[0] if guidance else None

    final_rows.append({
        "File": finding["file"],
        "Line": finding["line"],
        "Issue": finding["issue"],
        "Severity": finding["severity"],
        "Confidence": finding["confidence"],
        "Test_ID": finding["test_id"],
        "Security_Rule": best["title"] if best else "No matching rule",
        "Recommendation": best["recommendation"] if best else "No recommendation found",
        "Similarity": best["similarity"] if best else 0,
        "Average_Complexity": round(average_complexity, 2),
        "Risk_Score": risk_score,
        "Risk_Level": risk_level
    })

final_report = pd.DataFrame(final_rows)

display(final_report)
```

```python
# ============================================
# STEP 19: SAVE FINAL CSV
# ============================================

final_report.to_csv(
    "final_risk_report.csv",
    index=False
)

print("============================================")
print("FINAL REPORT CREATED SUCCESSFULLY")
print("============================================")
print("File: final_risk_report.csv")
print("Rows:", len(final_report))
print("Risk Score:", risk_score)
print("Risk Level:", risk_level)
```
